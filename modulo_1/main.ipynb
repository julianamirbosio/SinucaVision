```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
```

## Carregando imagem

```python
caminho_imagem = '/content/drive/MyDrive/Colab Notebooks/SinucaVision/data/img_1.jpg'

imagem = cv2.imread(caminho_imagem)

if imagem is None:
    print("Erro: Imagem não encontrada. Verifique o caminho!")
else:
    print("Imagem Original:")
    cv2_imshow(imagem)
```

## Isolando a mesa

```python
# 1. Converter BGR para HSV
hsv = cv2.cvtColor(imagem, cv2.COLOR_BGR2HSV)

# 2. Definir os limites do verde da mesa (ajuste conforme necessário)
# No OpenCV: H vai de 0-179, S vai de 0-255, V vai de 0-255
verde_escuro = np.array([40, 40, 40])
verde_claro = np.array([80, 255, 255])

# 3. Criar a máscara (tudo que for verde vira branco, o resto vira preto)
mascara_mesa = cv2.inRange(hsv, verde_escuro, verde_claro)

# 4. Aplicar a máscara na imagem original (para ver apenas a mesa)
resultado = cv2.bitwise_and(imagem, imagem, mask=mascara_mesa)

print("Máscara (Visão do Computador):")
cv2_imshow(mascara_mesa)

print("Resultado do Isolamento:")
cv2_imshow(resultado)
```
