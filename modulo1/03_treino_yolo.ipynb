## Célula 1 - Configuração e Treino
Vamos carregar o modelo base "nano" e apontar para o ficheiro `data.yaml`.

```python
from ultralytics import YOLO

modelo = YOLO('yolov8n.pt')
modelo.train(
    data='/content/dataset_aumentado/data.yaml',
    epochs=120,
    imgsz=640,
    batch=16,
    device=0,
    project='runs_sinuca',
    name='deteccao_bola'
)
```
